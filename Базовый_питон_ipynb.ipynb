```python
# | name | age |
# | ---- | NaN |
# | Anna | 23  |
# | Ivan | NaN |
# | Olga | 27  |

# Что сделать:
# -Найти пропущенные значения
# -Заменить NaN на средний возраст

import pandas as pd
data = [
    [pd.NA, pd.NA],
    ['Anna', 23],
    ['Ivan', pd.NA],
    ['Olga', 27]
]
columns = ['name', 'age']
df = pd.DataFrame(data = data, columns = columns)
print(df.head())

#Заменим NaN на средний возраст
total = 0 #сумма возрастов
count = 0 #счетчик (их количество)
nan_indexes = [] #список с запомненными индексами где лежат NaNы
for i in range(len(df)): #бежим по всем строчкам
    x = df.loc[i,'age'] #присваиваем переменной значение элемента в таблице (для удобства)
    if pd.notna(x): #проверяем, если он не NaN, то суммируем и добавляем счетчик
        total += x
        count += 1
    else:
        nan_indexes.append(i) #иначе запоминаем этот индекс
average = total / count #считаем среднее

for i in nan_indexes: #Заменяем пустые на среднее, доставая запомненные индексы
    df.loc[i, 'age'] = average #подставляем эти индексы в оригинальный датафрейм, чтоб оказаться на нужном месте и заменяем значение
print(df)

for i in range(len(df)): #та же логика для проверки имени, но вместо среднего значения поставим просто прочерк
    y = df.loc[i,'name']
    if pd.notna(y):
        continue
    else:
        df.loc[i,'name'] = '-'
print(f'\n Исправленная таблица \n {df}')
```

```python
#достань все числа в scientific notation
import re
text = "Values: 1.23E+3, -4.5E-2, 7E+10"
re.findall(r'[+-]?\d+(?:\.\d+)?[eE][+-]?\d+',text)

#[+-]? - есть ли какой то знак?
#\d+ - целое число
#(\.\d+)? - есть ли дробная часть (но так нельзя, тк питон вернет только то что в скобках)
#(?:\.\d+)? - есть ли дробная часть (без запоминания)
#[eE] - степень, важно! не \[eE] - это будет возвращением скобки [
#[+-]? - есть ли знак степени?
#\d+ - целое число
```
